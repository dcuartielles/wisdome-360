## The rule: measure type in degrees

On a dome, a letter's size is the **angle** it covers from the seats, not its size in pixels or points. **Cap height** is the height of a capital H.

> A 2048 master spans 180°, so **1° ≈ 11.4 px**. A 4096 master: **1° ≈ 22.8 px**.

| Cap height | Use | cap px at 2K | cap px at 4K | font size\* at 2K | real height on the dome |
|---|---|---|---|---|---|
| **8°** | titles | 91 | 182 | 131 | 82 cm |
| **6°** | titles (minimum) | 68 | 137 | 98 | 61 cm |
| **4°** | headings | 46 | 91 | 65 | 41 cm |
| **3°** | **comfortable body text** | 34 | 68 | 49 | 30 cm |
| **2°** | short labels | 23 | 46 | 33 | 20 cm |
| **1.5°** | **the minimum** | 17 | 34 | 24 | 15 cm |
| 1° | too small: letters built from about 11 pixels | 11 | 23 | 16 | 10 cm |
| 0.75° | too small: falls apart | 9 | 17 | 12 | 8 cm |

\*Font size for Poppins, whose cap height is 0.70 of the font size. For another typeface, measure its capital H.

**Real height** is the letter on the 11.6 m dome seen from the centre seat. From the back row everything looks about two-thirds the size, which is why 1.5° is a minimum, not a target.

## What breaks first

````python
det = tc.project(tc.detail(2048))
display(det.crop((424, 900, 1624, 1950)))
````

| Problem | What you see | Do instead |
|---|---|---|
| **Hairline weights** (Thin, ExtraLight) | strokes thinner than a pixel flicker, break up or vanish | Regular or heavier for anything under 3° |
| **Small serifs** | at 1.5° the serifs turn to mush and letters blur together | sans serif for small text; serifs only at title sizes |
| **Thin white on black** | thin bright strokes shimmer and show colour fringes on single-chip DLP projectors | soft white or a warm tone, regular or bold |
| **1–2 px lines** (of a 2K master) | they shimmer and disappear as the projectors resample | 3 px or more at 2K (6 px at 4K): about 0.3° |
| **Fine repeating stripes** | moiré and aliasing | nothing repeating finer than about 4 px at 2K |

## Why typed-straight-in text goes wrong

A dome master is a fisheye photograph of the dome, not a flat page. A straight line in the file is not a straight line in the room: it curves on the dome, more the further it is from the dome's zenith, which in Wisdome is 64° up in front of you. Letters also change size depending on where in the file they sit.

The cards use the [EX-08](https://dcuartielles.github.io/wisdome-360/examples/ex08-flat-to-dome/) method instead: one flat screen hanging in the room at the sweet spot, photographed through the 180° fisheye with the dome's 26° tilt. A flat screen magnifies slightly towards its top and bottom, so `type_card.py` corrects each row's font size for its height; the labelled size is exact where the row crosses the middle of the screen.

````python
flat = tc.project(tc.ladder(2048), flat=True)
display(flat.resize((1024, 1024)))
````

**For your own slides or titles:** design them flat, then put them through EX-08. On EX-08's default 60°-wide screen, 1.5° of cap height is about 4 % of the slide height (about 45 px on a 1080 px slide), and 3° is about 8 % (about 90 px).

## What usually breaks

| Symptom | Cause |
|---|---|
| `OSError: cannot open resource` | A font is missing from `fonts/` |
| Text that read on the laptop is mush in the dome | Measured in pixels on a monitor, not in degrees from the seats. Check it from the back row |
| Lines of text curve on the dome | Typed straight into the master. Put it on a screen in the room (EX-08) |
